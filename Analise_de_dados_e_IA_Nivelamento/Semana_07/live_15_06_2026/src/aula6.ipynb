```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
dados  = pd.read_csv(r'C:\Users\tlchaves\Documents\Santander\Santander-Imersao-Digital\Analise_de_dados_e_IA_Nivelamento\Semana_07\live_15_06_2026\db\dado_mg.csv')
```

```python
colunas = dados.columns[0]
colunas = colunas.split(';')
print(colunas)
dado = dados[dados.columns[0]][0].split(';')
dic = {colunas[i] : [dado[i]] for i in range(len(dado))}
for i in range(len(dados)):
  for key, value in zip(colunas,dados[dados.columns[0]][i].split(';')):
    dic[key].append(value)
dado = pd.DataFrame(dic)
dado.head()
```

```python
len(dados)
```

```python
dado['SG_UF_RESIDENCIA'].value_counts().sort_values(ascending=False).plot(kind='bar',logy = True,figsize = (20,10),title = "Quantidade de candidatos por estado",fontsize = 20)
plt.grid()
plt.show()
```
